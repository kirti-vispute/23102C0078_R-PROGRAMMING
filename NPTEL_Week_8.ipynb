## Lecture 32 — Vector Indexing

Vectors can be indexed with relational/logical conditions, the modulo operator, negative indices, names, and the empty index.

````r
x = 1:10
x
````

````r
x[ (x > 5) ]
````

````r
x[ (x%%2==0) ]  #%% indicates x mod y   #values for which x mod 2 is 0
````

````r
x[ (x%%2==1) ]  #values for which x mod 2 is 1
````

`NA` (missing values) can be inserted into a vector and filtered out with `is.na()`.

````r
x[5] = NA
x
````

````r
y = x[ !is.na(x) ]  #! Means negation
y
````

````r
mean(x)
````

````r
mean(y)
````

Negative indexing can also be used to exclude a range of positions.

````r
x = 1:10
x[-(1:5)]
````

````r
x[(6:10)]   # same outcome
````

### Named List/Vector Elements

Elements of a list can be given names, and a single name can be reassigned by indexing into `names()`.

````r
z = list(a1 = 1, a2 = "c", a3 = 1:3)
z
````

````r
names(z)
````

````r
names(z)[3] = "c2"
z
````

The same applies to vectors — elements can be named, and accessed by name.

````r
x = c(water=1, juice=2, lemonade=3 )
names(x)
````

````r
x["juice"]
````

### Empty Index

An empty index returns the whole vector.

````r
x = 1:10
x[]   # empty index returns whole vector
````

### Mixed-Mode List as a Matrix

Setting `dim()` on a list arranges its elements into a matrix-like structure, even when the elements are of different modes.

````r
ab = list(1, 2, 3, "X", "Y", "Z")
dim(ab) = c(2,3)
print(ab)
````

````r
mode(print(ab))
````

## Lecture 33 — Factors

Factors are R's way of representing categorical (qualitative) variables. A variable such as *Performance* (Excellent/Average/Good/Bad) can be coded numerically (1-4); a factor stores this coding together with the labels.

**Syntax**
```r
factor(x)
factor(x, levels)

factor(x = character(), levels, labels = levels, exclude = NA, ...)
```


````r
help("factor")
````

**Example:** factor of dice-roll outcomes, with the die faces given as labels instead of numbers 1-6.

````r
y = c(1, 4, 3, 5, 4, 2, 4)
possible.dieface = c(1, 2, 3, 4, 5, 6)
labels.dieface = c("one", "two", "three", "four", "five", "six")
facy = factor(y, levels = possible.dieface, labels = labels.dieface)
facy
````

## Lecture 34 — Factors: Class and Unclass

`as.factor()` converts a vector into a factor directly. For a **numeric** vector the levels are ordered numerically.

````r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
y = as.factor(x)
y
````

For a **character** vector, `factor()` orders the levels alphabetically by default.

````r
x = factor( c("lemonade", "lemonade", "juice", "lemonade", "water") )
x
````

### The `class()` Function

`class()` reports the class of an R object.

````r
class(9)
````

````r
class("9")
````

````r
class(print)
````

````r
x = matrix(nrow=2, ncol=2, data=1:4)
class(x)
````

````r
x
````

### The `unclass()` Function

`unclass()` strips the class attribute from an object, revealing its underlying representation — for a factor, the integer codes plus the `"levels"` attribute.

````r
brands = c("A","A","B","B","B","B","C")
brands_fac = factor(brands)
brands_fac
````

````r
unclass(brands_fac)
````

````r
colours = c("blue", "green", "red")
colours
````

````r
colours [unclass(brands_fac)]
````

````r
x = factor( c( "lemonade", "lemonade", "juice", "lemonade", "water") )
x
````

````r
unclass(x)
````

### Custom Factor Level Ordering with the `levels` Parameter

If a different assignment of levels is desired (instead of the default alphabetical order), the parameter `levels` can be used.

````r
x = factor( c("lemonade", "lemonade", "juice", "lemonade", "water"), levels=c("water", "juice", "lemonade") )
x
````

````r
unclass(x)
````

````r
levels(x)
````

### Ordered Factors

`ordered()` creates a factor whose levels have a defined order (e.g. low < medium < high).

````r
income = ordered(c("high", "high", "low", "medium", "medium"), levels=c("low", "medium", "high") )
income
````

````r
unclass(income)
````

## Lecture 35 — Strings: Display and Formatting ::: Print and Format Function

**Topics:** Formatting and Display of Strings; Operations with Strings.

We need formatting and display of strings to obtain the results of specific operations in the required format. The important commands regarding formatting and display are `print`, `format`, `cat`, and `paste`.

### The `print()` Function

`print()` prints its argument. It is a generic command that is available for every object class.

**Usage**
```r
print()
```


**Examples: Print numbers**

````r
print( sqrt(2) )
````

````r
print( sqrt(2), digits=5 )
````

````r
print( sqrt(2), digits=10 )
````

**Examples: Print characters**

````r
print("apple")
````

````r
print(c("apple", "banana"))
````

````r
print(c("apple", "banana", 6, 10))
````

### The `format()` Function

`format()` formats an R object for pretty printing.

**Usage**
```r
format(x, ...)
```
`x` is any R object; typically numeric.

```r
format(x, trim = FALSE, digits = NULL, nsmall = 0L,
       justify = c("left", "right", "centre", "none"),
       width = NULL, ...)
```
- `digits` — how many significant digits are to be used
- `nsmall` — the minimum number of digits to the right of the decimal point
- `justify` — left-justified (the default), right-justified, or centred


````r
print( format( 0.5, digits=10, nsmall=15 ) )
````

### Formatting and Display of Strings: Use of `width`

````r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
````

````r
format(c("A", "BB", "CCC", "DDDD"), width = 14, justify = "centre")
````

### Formatting and Display of Strings: Use of `justify`

````r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "left")
````

````r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "right")
````

````r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "none")
````

### Formatting a Matrix

A matrix is displayed in the R command window; the desired number of digits can be specified with the option `digits`.

````r
x = matrix(nrow=3, ncol=2, data=1:6, byrow=T)
print(x)
````

### Formatting and Display of Large Quantities

The `big.mark` command (used with `format`) prints large quantities with sequences separated by a chosen character, e.g. a comma.

````r
format(1234567, big.mark = ",")
````

````r
format(12345678, big.mark = ",")
````

````r
format(123456789, big.mark = ",")
````

Sequences can also be separated by spaces instead of commas.

````r
format(123456789, big.mark = " ")
````
