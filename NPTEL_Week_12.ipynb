## Lecture 50 — Graphics: Sub-Divided Bar Plots and Pie Diagram

### Subdivided (component) bar diagram
Subdivided or component bar diagram divides the total magnitude of variables into various parts.

**Example:** The data on the number of customers visiting 3 shops during 10-11 AM on 4 consecutive days.

````r
cust = matrix(nrow=4, ncol=3, data =c(2,20,
30,26,53,40,42,15,25,30,75,100), byrow = T)

cust
````

**Usage:** `barplot(variable in matrix format)` will create a subdivided or component bar diagram with columns of matrix as bars. Sections inside bars indicate the values in cumulative form.

````r
barplot(cust)
````

**Adding labels and colours:**

````r
barplot(cust, names.arg=c("Shop 1", "Shop 2", "Shop 3"),
        xlab = "Shops", ylab = "Days",
        col = c("red","green","orange","brown"))
````

### Pie diagram
Pie charts visualize the absolute and relative frequencies. A pie chart is a circle partitioned into segments where each segment represents a category. The size of each segment depends upon the relative frequency and is determined by the angle (frequency X 360°).

Usage: `pie(x, labels = names(x), ...)`

**Example:** Code the 10 persons by using, say 1 for male (M) and 2 for female (F).

`M, F, M, F, M, M, M, F, M, M` → `1, 2, 1, 2, 1, 1, 1, 2, 1, 1`

````r
gender = c(1, 2, 1, 2, 1, 1, 1, 2, 1, 1)

gender

pie(gender)
````

A pie chart of the *raw* vector gives one slice per observation. Usually you want the **frequency table** instead:

````r
pie(table(gender))
````

### Combining graphics
Use command `par()` to put multiple graphs in a single plot. Adjust the graphical parameters with the help of function.

`par(mfrow=c(p,q))` sets the plotting area into a p*q array.

````r
par(mfrow=c(1,2))    # set the plotting area into a 1*2 array

barplot(table(direction))

pie(table(direction))
````

## Lecture 51 — Graphics: Histogram

Histogram is based on the idea to categorize the data into different groups and plot the bars for each category with height. Data is continuous. The area of the bars (= height X width) is proportional to the frequency (or relative frequency), so the widths of the bars need not necessarily be the same.

```r
hist(x)            # show absolute frequencies
hist(x, freq=F)     # show relative frequencies
```

Full signature: `hist(x, main, col, xlab, xlim, ylim)` — see `help("hist")` for more details.

**Example:** Height of 50 persons in centimetres are recorded as follows:

````r
height = c(166,125,130,142,147,159,159,147,
165,156,149,164,137,166,135,142,133,136,127,143,
165,121,142,148,158,146,154,157,124,125,158,159,
164,143,154,152,141,164,131,152,152,161,143,143,
139,131,125,145,140,163)
````

````r
hist(height)
````

**Adding colour to bars and titles on axis:**

````r
hist(height, main = "Heights of persons", col = "green",
     xlab = "Heights", ylab = "Number of Persons")
````

**Adding density** — `density`: density of shading lines, in lines per inch. Non-positive values of density also inhibit the drawing of shading lines.

````r
hist(height, main = "Heights of persons", col = "red",
     xlab = "Heights", ylab = "Number of Persons", density = 2)
````

**Increasing density:**

````r
hist(height, main = "Heights of persons", col = "red",
     xlab = "Heights", ylab = "Number of Persons", density = 8)
````

**Adding angle** — `angle`: the slope of shading lines, given as an angle in degrees (counter-clockwise).

````r
hist(height, main = "Heights of persons", col = "red",
     xlab = "Heights", ylab = "Number of Persons", density = 8, angle = 100)
````

## Lecture 52 — Graphics: Bivariate and Three Dimensional Scatter Plots

### Association of two variables
How to judge or graphically summarize the association of two variables? Graphical procedures include two dimensional plots and three dimensional plots; quantitative procedures include correlation coefficients, contingency tables, Chi-square statistic, linear/nonlinear regression etc.

### Bivariate scatter plots
Plot command:
```r
plot(x, y)
plot(x, y, type)
```
`type` can be `"p"` for points, `"l"` for lines, `"b"` for both, `"c"` for the lines part alone of `"b"`, `"o"` for both 'overplotted', `"s"` for stair steps, `"h"` for 'histogram' like (or 'high-density') vertical lines.

Other options: `main` (overall title), `sub` (subtitle), `xlab`/`ylab` (axis titles), `asp` (y/x aspect ratio). See `help("type")` for more details.

**Example:** Data on marks obtained by 20 students out of 500 marks and the number of hours they studied per week.

````r
marks = c(337,316,327,340,374,330,352,353,370,380,
          384,398,413,428,430,438,439,479,460,450)

hours = c(23,25,26,27,30,26,29,32,33,34,35,38,39,42,
          43,44,45,46,44,41)
````

````r
plot(hours, marks)
````

`"l"` for lines:

````r
plot(hours, marks, "l")
````

`"b"` for both -- line and point:

````r
plot(hours, marks, "b")
````

`"o"` for both 'overplotted':

````r
plot(hours, marks, "o")
````

`"h"` for 'histogram' like (or 'high-density') vertical lines:

````r
plot(hours, marks, "h")
````

`"s"` for stair steps:

````r
plot(hours, marks, "s")
````

**Adding axis labels and a title:**

````r
plot(hours, marks, xlab = "Number of weekly hours",
     ylab = "Marks obtained",
     main = "Marks obtained versus Number of hours per week")
````

### Scatter plots with a smooth curve
`scatter.smooth` produces a scatter plot and adds a smooth curve, based on LOESS (locally weighted scatterplot smoothing), used for local polynomial regression fitting.

```r
scatter.smooth(x, y = NULL, span = 2/3, degree = 1,
               family = c("symmetric","gaussian"), xlab = NULL,
               ylab = NULL, ylim = range(y, pred$y, na.rm = TRUE), ...)
```
- `x, y`: the x and y coordinates for the plot.
- `span`: smoothness parameter for LOESS.
- `degree`: degree of local polynomial used.
- `family`: if `"gaussian"` fitting is by least-squares, and if `"symmetric"` a re-descending M estimator is used.
- `xlab`, `ylab`: axis labels. `ylim`: the y limits of the plot.

See `help("scatter.smooth")` for more details.

````r
scatter.smooth(hours, marks)
````

**Adding more options:**

````r
scatter.smooth(hours, marks, lpars = list(col =
"red", lwd = 3, lty = 3))
````

## Lecture 53 — Some Examples of R Programming

### Steps to write a programme
A programme is a set of instructions or commands written in a sequence of operations. Programme in R is written as a function using `function`. Write down the objective, identify the input and output variables and their nature (numeric, string, factor, matrix etc.).

**Tips:**
- Loops usually slow down programmes, so it is often better to use vectors and matrices.
- Use the `#` symbol to write comments to understand the syntax.
- Use variable names which are easy to understand.
- Don't forget to initialize the variables.

### Example 1
Suppose we want to compute

$$g = \dfrac{\sum_{i=1}^n x_i^2}{\sum_{i=1}^n y_i^2} \qquad \text{and} \qquad h = \sum_{i=1}^n \left(\dfrac{x_i}{y_i}\right)^2$$

**Input variables:** `x`, `y`, `n` (if `x` and `y` have different number of observations, choose different numbers, say `n1` and `n2`)

**Output variables:** `g`, `h`

We need summation, so use the `sum` function, or alternatively compute it through vectors.

````r
# Remove all data
rm(list = ls())

# Define input data vectors, for example
x = c(10,20,30)
y = c(1,2,3)

######START OF FUNCTION########
example1 = function(x,y)

# Start of function body
{
# First give all other input variables

# Computation of number of observations
n = length(x)

#Initialize the values to store squared values
x1 = 0
y1 = 0
z1 = 0

#Start of loop
for (i in 1:n)
{
# Define x1, y1 and z1 to store their squares
  x1[i] = x[i]^2
  y1[i] = y[i]^2
  z1[i] = (x[i]/y[i])^2
#End of loop
  }

# Obtain the sum of squared quantities
sum_square_x = sum(x1)
sum_square_y = sum(y1)
sum_square_z = sum(z1)

# Computation of g and h
g = sum_square_x/sum_square_y
h = sum_square_z

# Format the output
cat("The value of g and h are", g, "and", h,
"respectively", "\n")
}
######END OF FUNCTION########
````

**Trying it out** — just by changing the values of `x` and `y`, one can get different outcomes:

````r
x=c(10,20,30)
y=c(1,2,3)
example1(x,y)
# The value of g and h are 100 and 300 respectively

x=c(67,87,26,85,6,45)
y=c(54,64,22,94,20,88)
example1(x,y)
# The value of g and h are 0.8996568 and 5.953203 respectively
````

**Example 1 (alternative, vectorised approach)** — avoids the explicit loop entirely:

````r
g = sum(x^2)/sum(y^2)
h = sum(x/y)^2
````

### Example 2
Suppose we want to compute

$$f(x,y) = \dfrac{\left(\dfrac{x+\ln y}{y}\right)^2}{5+\left(\dfrac{x+\ln y}{y}\right)^3}\left[\exp\left(\dfrac{x+\ln y}{y}\right)\right]^{2/3}$$

This can be written as $f(x,y) = \dfrac{(g(x,y))^2}{5+(g(x,y))^3}\left[\exp(g(x,y))\right]^{2/3}$ where $g(x,y) = \dfrac{x+\ln y}{y}$.

**Input variables:** `x`, `y`. **Output variable:** `f`.

We break this function into two components -- compute `g(x,y)` as a function, and then compute `f(x,y)` by calling `g(x,y)`.

````r
# Remove all data
rm(list = ls())

# Define input data vectors x, y
# (left as placeholders here -- concrete values are assigned in the
#  next cell before f(x,y) is called)

# define g(x,y)
g = function(x,y)
# Start of function
{
   (x+log(y))/y
# End of function
}

++++++++++++++++++++++++++++++++++++++++

# define f(x,y)
f = function(x,y)
{
(((g(x,y))^2)/(5+(g(x,y))^3))*(exp(g(x,y)))^(2/3)
}
# g(x,y) must have been defined earlier.
````

**Trying it out** — there is no need to calculate the value of `g(x,y)` separately; `f` calls it internally:

````r
x=10
y=20
f(x,y)
# [1] 0.1234539

x=1896
y=23454
f(x,y)
# [1] 0.001394291
````

### Example 3
Suppose we want to compute the piecewise function

$$f(x) = \begin{cases} \exp\left(\dfrac{x+\ln(1+x^3)}{x^2}\right) & \text{if } x>0 \\ 10 & \text{if } x=0 \\ \dfrac{2+x^3}{x} & \text{if } x<0 \end{cases}$$

and plot it with a line over values of `x` as a sequence starting from -1 to 5 and increasing it by 0.2.

**Input variable:** `x`. **Output variable:** `f`.

````r
# Remove all data
rm(list = ls())

# Define input data x
# (left as a placeholder here -- concrete values are passed to f(x)
#  directly in the test calls a couple of cells below)

f = function(x)
{
  if(x>0) {exp((x+log(1+x^3))/x^2)}

  else if(x==0) {10}

  else {(2+x^3)/x}
}

h = function()
# Start of function
{
# Generation of data on x
x = seq(-1,5,by=0.2)
# Initialization of y to store values of f(x)
y = 0

# Generation of f(x) values corresponding to x
for(i in 1:length(x))
{
  y[i] = f(x[i])
}
# length(x) and length(y) must be same to plot
# y = f(x) with respect to x
plot(x, y, type = "l")
}
````

**Trying `f` on single values:**

````r
f(123)
# [1] 1.009126

f(-123)
# [1] 15128.98

f(0)
# [1] 10

f(8)
# [1] 1.249201

f(-4)
# [1] 15.5

f(0)
# [1] 10
````

**Generating the full plot over the sequence of `x` values:**

````r
h()
````
