## Lecture 22 — Loops: the `for` loop

**Control structures in R:** Control statements, Functions, Loops.

**Loops** execute repetitive commands: `for` loop, `while` loop, `repeat` loop.

### 1. The `for` loop

If the number of repetitions is known in advance (e.g. if all commands have to be executed for all cases *i* = 1,2,...,*n* in the data), a `for()` loop can be used.

**Syntax**
```r
for (name in vector) {commands to be executed}
```
A variable with name `name` is sequentially set to all values contained in the vector `vector`. All operations/commands are executed for all these values.


````r
## Example 1
for ( i in 1:5 ) { print( i^2 ) }
````

*(`print` is a function to print the argument.)*

````r
## Example 2
for ( i in c(2,4,6,7) ) { print( i^2 ) }
````

````r
## Example 3
x = c(2,4,6,8,10,12)

excount = function(x){
count = 0
for (xval in x) {
if(xval/2 > 3)
count = count+1
}
print(count)
}
excount(x)
````

### Nested looping with `for` loop

We can have a loop inside a loop.

````r
## Example 4
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
    for (y in sweet) {
        print(paste(x, y))
    }
}
````

### The `break` command

Using the `break` command, we can stop the loop before it has looped through all the items.

````r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
````

### The `next` command

Using the `next` command, we can skip an iteration without terminating the loop.

Suppose we want to skip lemonade.

````r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
````

Suppose we want to skip tea.

````r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
````

## Lecture 23 — Loops: `while` and `repeat`

### 2. The `while` loop

If the number of loops is not known before (e.g. when an iterative algorithm to maximize a likelihood function is used), one can use a `while()` loop.

**Syntax**
```r
while(condition){ commands to be executed as long as condition is TRUE }
```
If the condition is not true *before entering* the loop, no commands within the loop are executed.


````r
## Example 1
i = 1
while (i < 10) {
print(i^2)
i = i+2
}
````

### 3. The `repeat` loop

The `repeat` loop doesn't test any condition — in contrast to the `while()` loop — before entering the loop and also not during the execution of the loop.

Again, the programmer is responsible that the loop terminates after the appropriate number of iterations. For this the `break` command can be used.

**Syntax**
```r
repeat{ commands to be executed }
```


````r
## Example 3
i = 1
repeat{
print( i^2 )
i = i+2
if ( i > 10 )
break
}
````

### Example 4

Additionally, the command `next` is available, to return to the beginning of the loop (to return to the first command in the loop).

````r
i = 1
repeat{
i = i+1
if (i < 10) next
print(i^2)
if (i >= 13) break
}
````

## Lecture 24 — Functions

Functions are a bunch of commands grouped together in a sensible unit. Functions take input arguments, do calculations (or make some graphics, call other functions) and produce some output and return a result in a variable. The returned variable can be a complex construct, like a list.

**Components:**
- **Function Name** — the name of the function, which is stored as an object with this name.
- **Arguments** — contain the input values. A function may contain no arguments.
- **Function Body** — contains statements that define what the function has to do.
- **Return Value** — the evaluated value of the last expression in the function body.

**Built-in functions:** R has built-in functions also. Simple examples are `sum()`, `prod()`, `mean()`, `max()`, `sum(x)`, etc.

**Syntax**
```r
Name <- function(Argument1, Argument2, ...)
        {
         expression(s)
                    }
```
where `expression(s)` is a single command or a group of commands. You can use the `=` operator also:
```r
Name = function(Argument1, Argument2, ...)
        {
         expression(s)
                    }
```
Function arguments can be given a meaningful name, can be set to default values, and functions can have the special argument `'...'`.


````r
## Functions (single variable)
abc = function(x){
            x^2
             }
abc
abc(3)
abc(6)
abc(9)
````

````r
## Functions (two variables)
abc = function(x,y){
          x^2+y^2
              }
abc(3,4)
abc(10, 10)
abc(-2, -3)
````

````r
## Functions - another example
abc = function(x){
        sin(x)^2+cos(x)^2 + x
                }
abc(9)
abc(99)
abc(-15)
````

### Calling a function without an argument

````r
abc = function() {
    for(i in 1:3) {
        print(i^3)
    }
}

abc()
````

## Lecture 25 — Sequences

A sequence is a set of related numbers, events, movements, or items that follow each other in a particular order. The regular sequences can be generated in R.

**Syntax**
```r
seq()

seq(from = 1, to = 1,
by = ((to - from)/(length.out - 1)),
length.out = NULL, along.with = NULL, ...)
```
Help for `seq`:
```r
help("seq")
```


````r
help("seq")
````

The default increment is +1 or −1.

````r
seq(from=2, to=4)
````

````r
seq(from=4, to=2)
````

````r
seq(from=-4, to=4)
````

**Sequence with constant increment:** generate a sequence from 10 to 20 with an increment of 2 units.

````r
seq(from=10, to=20, by=2)
````

**Sequence with constant decrement:** generate a sequence from 20 to 10 with a decrement of 2 units.

````r
seq(from=20, to=10, by=-2)
````

**Downstream sequence with constant decrement:** generate a sequence from 3 to -2 with a decrement of 0.5 units.

````r
seq(from=3, to=-2, by=-0.5)
````

**Sequences with a predefined length with default increment +1**

````r
seq(to=10, length=10)
````

````r
seq(from=10, length=10)
````

**Sequences with a predefined length with constant fractional increment**

````r
seq(from=10, length=10, by=0.1)
````

**Sequences with a predefined length with constant decrement**

````r
seq(from=10, length=10, by=-2)
````

**Sequences with a predefined length with constant fractional decrement**

````r
seq(from=10, length=5, by=-.2)
````
