## Lecture 16 — Logical Operators

### Logical Operators and Comparisons

| Operator | Executions |
|---|---|
| `>` | Greater than |
| `>=` | Greater than or equal |
| `<` | Less than |
| `<=` | Less than or equal |
| `==` | Exactly equal to |
| `!=` | Not equal to |
| `!` | Negation (not) |
| `&`, `&&` | and |
| `\|`, `\|\|` | or |
| `xor(x,y)` | either... or (exclusive) |
| `isTRUE(x)` | test if x is TRUE |
| `isFALSE(x)` | test if x is FALSE |
| `TRUE` / `FALSE` | true / false |

The **shorter form** (`&`, `|`) performs element-wise comparisons in almost the same way as arithmetic operators. The **longer form** (`&&`, `||`) evaluates left to right, examining only the first element of each vector — evaluation proceeds only until the result is determined.

### Examples: Use of `|` and `||`

```r
x=8
(x < 10) || (x < 2)
```

```r
x=18
(x < 10) || (x < 2)
```

`||` operates only on the first element and not on remaining elements; `|` operates on all the elements.

> **Note on R version:** the notes were recorded on an older R version, where `||`/`&&` on a length>1 vector silently used just the first element (with a warning). Since R 4.3.0, this is a hard **error** instead — `'length = 2' in coercion to 'logical(1)'`. The cell below is wrapped in `tryCatch()` so the real, current behavior is shown rather than hidden or faked.

```r
x = c(8,18)
tryCatch(
  (x < 10) || (x < 2),
  error = function(e) e
)
```

```r
(x < 10) | (x < 2)
```

### Examples: Use of `&` and `&&`

```r
x = 5
(x < 10) && (x > 2)
```

```r
x = 15
(x < 10) && (x > 2)
```

`&&` operates only on the first element and not on remaining elements; `&` operates on all the elements.

> **Note on R version:** as above, `&&` on a length>1 vector is a hard error in this R version (4.3.3) rather than the old silent first-element behavior, so the cell below is wrapped in `tryCatch()` to show the real result.

```r
x = c(8,18)
tryCatch(
  (x < 10) && (x > 2),
  error = function(e) e
)
```

```r
(x < 10) & (x > 2)
```

### Examples

```r
x = 1:6        # Generates x=1,2,3,4,5,6
x
```

```r
(x > 2) & (x < 5)  # Checks whether the values are greater than 2 and less than 5
```

```r
x[(x > 2) & (x < 5)] # Finds which values are greater than 2 and smaller than 5
```

```r
(x > 2) | (x < 5)  # Checks whether the values are greater than 2 or less than 5
```

```r
x[(x > 2) | (x < 5)] # Finds which values are greater than 2 or smaller than 5
```

### Example of "The Longer Form Evaluates Left to Right Examining Only the First Element of Each Vector"

> **Note on R version:** the notes show `(x > 2) && (x < 5)` on a length-6 `x` silently evaluating as if only the first element were used, which was R's old behavior. In R 4.3.3 this is a hard error instead (`'length = 6' in coercion to 'logical(1)'`), so the cell below uses `tryCatch()` to show the real current result, immediately followed by the single-element version the notes say it is equivalent to.

```r
x = 1:6        # Generates x = 1,2,3,4,5,6
tryCatch(
  (x > 2) && (x < 5),
  error = function(e) e
)
```

is equivalent to:

```r
(x[1] > 2) & (x[1] < 5)   # Note that x[1] is only the first element in x
```

## Lecture 17 — Relational and Logical Operators

### Relational Operators and Comparisons

`TRUE` and `FALSE` are reserved words denoting logical constants. Zero is considered as `FALSE` and non-zero numbers are taken as `TRUE`.

| Operator | Executions |
|---|---|
| `>` | Greater than |
| `>=` | Greater than or equal |
| `<` | Less than |
| `<=` | Less than or equal |
| `==` | Exactly equal to |
| `!=` | Not equal to |
| `!` | Negation (not) |

### Logical Operators and Comparisons

| Operator | Executions | |
|---|---|---|
| `!` | Logical not | |
| `&` | Element-wise logical and | Performs operation element-wise |
| `&&` | Logical and | examines only the first element of the operands |
| `\|` | Element-wise logical or | Performs operation element-wise |
| `\|\|` | Logical or | examines only the first element of the operands |
| `xor(x,y)` | Either x or y (x or y but not x and y) | |
| `isTRUE(x)` | test if x is TRUE | |
| `isFALSE(x)` | test if x is FALSE | |
| `TRUE` / `FALSE` | true / false | |

### Example of Standard Logical Operations

Truth table — Statement 1 (x), Statement 2 (y), Outcome (x and y), Outcome (x or y):

| x | y | x and y | x or y |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |


```r
x = TRUE
y = FALSE
```

```r
x & y     # x AND y
```

```r
x | y     # x OR y
```

```r
!x         # negation of x
```
