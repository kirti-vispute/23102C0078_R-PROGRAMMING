```python
# IF ELSE CONTROL STRUCTURE

%%R
x = 5
if ( x==3 ) { x = x-1 } else { x = 2*x }
x
```

```python
%%R
x = 3
if ( x==3 ) { x = x-1 } else { x = 2*x }
x
```

```python
%%R
x = 6
if(x > 3){
print("The value is more than 3")
} else {
print("The value is less than 3")
}
```

```python
%%R
x = 2
if(x > 3){
print("The value is more than 3")
} else {
print("The value is less than 3")
}
```

```python
# NESTED IF ELSE CONTROL STRUCTURE

%%R
x = 5
if ( x==3 ) {
x = x-1
} else if ( x < 3 ) {
x = x+5
} else { x = 2*x }
x
```

```python
%%R
x = 2
if ( x==3 ) {
x = x-1
} else if ( x < 3 ) {
x = x+5
} else { x = 2*x }
x
```

```python
%%R

x = 3
if ( x==3 ) {
x = x-1
} else if ( x < 3 ) {
x = x+5
} else { x = 2*x }
x
```

```python
# IF ELSE CONDITIONAL EXCECUTION

%%R
x = 1:10
x
ifelse( x<6, x^2, x+1 )
```

```python
%%R
x = c(7,9,8,4)
ifelse(x %% 2 == 0,"even number","odd number")
```
